## Reading the processed data into a dataframe

Within the CSV file, there is an untitled index. Reading the CSV file normally turns this index into a column titled "Unnamed: 0", and creates a new index. So, I have to register this column as the index of the dataframe by using `index_col="Unnamed: 0"`.

```python
processed_data_path = "data/processed/processed_spotify_data.csv"
processed_data_df = pd.read_csv(processed_data_path, index_col="Unnamed: 0")
```

## Viewing the dataframe

```python
processed_data_df.head()
```

## Randomly sampling the dataframe

Here, I take a random sample of 1000 rows of the full processed date. Then, I make sure the raws are in ascending order of index. Finally, I reset the indices so they run from 0 to 999.

```python
sample_df = processed_data_df.sample(n=1000, random_state=1)
sample_df = sample_df.sort_index()
sample_df = sample_df.reset_index(drop=True)
sample_df.head(5)
```

## Writing processed data sample to CSV file

```python
processed_location = "data/processed/processed_data_sample.csv"
sample_df.to_csv(processed_location, encoding="utf-8")
```
